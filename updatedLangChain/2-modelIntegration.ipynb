GROQ MODEL INTEGRATION


```python
import os
from langchain.chat_models import init_chat_model
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
model = init_chat_model("groq:openai/gpt-oss-120b")
response = model.invoke("how are you doing today?")
response.content
```

```python
model.invoke("why i am so busy today?")
```

problem : in model.invoke(""): we have to wait
solution : Streaming
Most LLM models can stream content while generating their output it improves use experience when output is too large

```python
model.stream("right a paragraph about the importance of AI in healthcare.200 words")
```

```python
for chunk in model.stream("right me 2000 word paragraph in artificial intelligence"):
 print(chunk.content, end="|",flush=True)
```

Batching:
Sending multiple independent request to llm together so it can process all request parallely 

```python
response=model.batch(["what is the capital of France?", "what is the capital of Germany?"])
for res in response:
    print(res.content)
```

```python
response=model.batch(["what is the capital of France?", "what is the capital of Germany?"],config={'max_concurrency': 5})
for res in response:
    print(res.content)
```
