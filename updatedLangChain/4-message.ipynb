Messages
Messages are fundamental unit of context for models in langchain. They represent input and output of model carrying both content and metadata needed to represnet state of converion while interacting with llm. Messgaes are object that contains
1. Role: identify message type (user,system)
2.content: represnet actual content of the messgae
3. metadata: option fields like token usage,messgae Ids,response information

```python
import os
from dotenv import load_dotenv  
from langchain.chat_models import init_chat_model
load_dotenv()
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
model = init_chat_model("groq:openai/gpt-oss-120b")

```

Text Prompts
text prompts are string. Ideal for straight forward generation task where we don't need to retain conversation history 

use text prompts where
we don't want any conversation history
for simple task where we want less complexity
when have single standalone request

Message Prompts:
we can pass a list of mesaage to modle by passing a list of message objects

Message types
1. system message: tells the model how to behave and provide context for interation
2. Human message : represent user input and interaction with model
3. tool message: represent output of tool calls
4. AI message : response generated by model contains content,metadata,tool calls


```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage, ToolMessage
messages=[
    SystemMessage(content="You are a American politics language teacher. You are very good at answering questions"),
    HumanMessage(content="Who is president?"),
    # AIMessage(content="The prime minister of India is Narendra Modi.")
    ]
conversation_messages=[SystemMessage("you are a helpful assistant"), 
                       HumanMessage("will you help me"),
                       AIMessage("yes I will help you"),
                       HumanMessage("when Tajmahal was built?"),]
response=model.invoke(conversation_messages)
print(response.content)
```
